### Q11. Linear algebra with NumPy
Solve the following system of equations using `np.linalg.solve()`:

2x + y = 11

x + 3y = 18

Print the values of x and y.

```python
import numpy as np

A = np.array([[2, 1],
              [1, 3]])

B = np.array([11, 18])

x, y = np.linalg.solve(A, B)

print("x =", x)
print("y =", y)
```

### Q13. Why NumPy is faster than plain Python
Create two Python lists of 500,000 numbers each. Time how long it takes to add them element-by-element using a Python loop (or list comprehension). Then convert both lists to NumPy arrays and time how long it takes to add them using `+`. Print both times and, in a comment, state which was faster and roughly by how much.

```python
import numpy as np
import time

list1 = list(range(500000))
list2 = list(range(500000))

start = time.time()
result_python = [a + b for a, b in zip(list1, list2)]
python_time = time.time() - start

arr1 = np.array(list1)
arr2 = np.array(list2)

start = time.time()
result_numpy = arr1 + arr2
numpy_time = time.time() - start

print("Python list time:", python_time, "seconds")
print("NumPy array time:", numpy_time, "seconds")

if numpy_time > 0:
    print("NumPy was roughly", python_time / numpy_time, "times faster.")
```

### Q15. Indexing and slicing to build X and y
Create a 2-D array with 6 rows and 3 columns, where the last column is a target label (0 or 1) and the first two columns are features (make up your own numbers).
1. Print the first 3 rows only.
2. Print only the first column, for all rows.
3. Split the array into `X` (all columns except the last) and `y` (only the last column), and print both.

```python
import numpy as np

data = np.array([
    [10, 20, 0],
    [15, 25, 1],
    [20, 30, 0],
    [25, 35, 1],
    [30, 40, 1],
    [35, 45, 0]
])

print("First 3 rows:")
print(data[:3])

print("\nFirst column:")
print(data[:, 0])

X = data[:, :-1]
y = data[:, -1]

print("\nX (features):")
print(X)

print("\ny (target):")
print(y)
```

### Q16. Vectorized operations and boolean masking
Create a 1-D NumPy array of 8 exam scores (0-100) of your choice. Without using any loop:
1. Add 5 bonus marks to every score and print the result.
2. Print a boolean array showing which scores are 50 or above.
3. Print only the scores that are 50 or above using boolean masking.

```python
import numpy as np

scores = np.array([45, 67, 82, 39, 55, 91, 48, 73])

bonus_scores = scores + 5
print("Scores after bonus:", bonus_scores)

passed = bonus_scores >= 50
print("Boolean array:", passed)

print("Scores 50 or above:", bonus_scores[passed])
```
